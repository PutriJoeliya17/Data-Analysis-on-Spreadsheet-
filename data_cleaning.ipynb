```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('portugal_listinigs.csv', low_memory=False)
df
```

#### Memuat Informasi Data Mentah Sebelum di Cleaning

```python
df.info()
```

```python
print(len(df))
```

#### Membersihkan Data yang Duplikat

```python
df = df.drop_duplicates()
```

#### Mengkerucutkan Fokus Hanya Kepada Perumahan Saja 

```python
df = df[df['Type'].isin(['Apartment', 'House'])]
```

#### Membersikah Outliers

membuang semua properti yang berada di area yang tidak masuk akal dan juga jenis properti yang memiliki ukuran yang tidak masuk akal untuk di perjual belikan. 

```python
df = df[(df['TotalArea'] >= 5) & (df['TotalArea'] <= 1500)]
```

#### Membersihkan Harga yang Tidak Realistis 

membuang semua properti dengan harga di bawah 5000 Euro karena tidak masuk akal. 

```python
df = df[df['Price'] > 5000]
```

#### Mengisi Missing Value Dengan Bool 

pada kolom kolom tersebut di pilih menggunakan boolean karena hanya memerlukan keteranyan apakah tersedia atau tidak deimana jika tersedia fasilitas tersebut maka keterangkan yang muncul adalah "True"

```python
df['Garage'] = df['Garage'].fillna(False)
df['Elevator'] = df['Elevator'].fillna(False)
df['ElectricCarsCharging'] = df['ElectricCarsCharging'].fillna(False)
```

#### Mengisi Missing Value Dengan Keterangan Lain 

pada kolom kolom tersebut di di isi keterangan pada missing value dengan keterangan "Unkown" atau "NC" untuk menjelaskan bahwa membah kotak data tersebut tidak diisi karna kosong atau tidak tersedia 

```python
df['ConservationStatus'] = df['ConservationStatus'].fillna('Unknown')
df['EnergyCertificate'] = df['EnergyCertificate'].fillna('NC')
```

#### Menghapus Semua Baris Kolom Yang Kosong 

fungsi ini dilakukan pada kolom berguna untuk bisa membuang semua baris yang tidak memiliki keterangan price dan total area karna hal ini di rasa penting untuk bisa tersedia dan di analisis, selain ini kolom ini penting untuk para calon penbeli sehingga di butuhkan angka yang pasti.

```python
df = df.dropna(subset=['Price', 'TotalArea'])
```

#### Menghapus Barisan yang tidak Memiliki Kamar mandi 

di dalam data terdapat beberapa rumah dan apartemen yang tidak memiliki kamar mandi sehingga dirasa hal ini tidak masuk akal dan perlu di hapus saja 

```python
df = df[df['NumberOfBathrooms'] > 0]
```

#### Mengisi Missing Value Dengan Median 

Jumlah kamar yang datanya hilang di isi dengan median untuk sebagai nilai tengah 

```python
df['NumberOfBedrooms'] = df['NumberOfBedrooms'].fillna(df['NumberOfBedrooms'].median())
```

#### Menghapus Row yang Tidak Memiliki Publish Date

dilakukan untuk bisa menganalisis trend pasar yang ada, walaupun datanya berkurang jauh tetapi penting untuk bisa menyajikan data dengan quality yang baik di banding quantity

```python
df = df.dropna(subset=['PublishDate'])
```

#### Membuat Tabel Final yang Berisikian Kolom Kolom yang Hanya Digunakan Untuk di Analisis

```python
kolom_diikuti = ['Price', 'District', 'City', 'Town', 'Type', 'EnergyCertificate', 
                 'NumberOfBathrooms', 'TotalArea', 'PublishDate', 'Garage', 
                 'Elevator', 'NumberOfBedrooms', 'ConservationStatus']
df_final = df[kolom_diikuti]
```
